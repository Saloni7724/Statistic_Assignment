```python
""" 
1.Generate 100 random numbers in Python that follow a normal distribution with mean 50 and standard deviation 10 using numpy, then plot the bell curve
using matplotlib.
"""

import numpy as np
import matplotlib.pyplot as plt

# Generate 100 random numbers with mean 50 and standard deviation 10
data = np.random.normal(50, 10, 100)

# Plot histogram
plt.hist(data, bins=10, density=True, alpha=0.7)

plt.title("Normal Distribution - Mean 50, Standard Deviation 10")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
""" 
2.Given an array of 20 daily step counts from your fitness tracker app, calculate the mean,median, and mode, and check if they are approximately equal
 as expected in a normal distribution.<br><br><em><strong>Hint:</strong> Use numpy.mean(), numpy.median(), and scipy.stats.mode().</em>
 
 """
import numpy as np
from scipy import stats

# Daily step counts for 20 days
steps = [
    8000, 8200, 7800, 8500, 8100,
    7900, 8300, 8000, 8400, 8100,
    8200, 8000, 7900, 8300, 8100,
    8000, 8200, 8100, 8400, 8000
]

mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps, keepdims=True).mode[0]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)

if abs(mean - median) < 300 and abs(median - mode) < 300:
    print("Mean, median and mode are approximately equal.")
else:
    print("Mean, median and mode are not approximately equal.")
```

```python
"""  
3.Simulate the scores of 200 players in an online game (e.g., PUBG, Free Fire) using a normal distribution with mean 70 and standard deviation 15, 
then count how many players scored within one standard deviation (between 55 and 85) of the mean.<br><br><em> <strong>Hint:</strong> Use the 68-95-99 rule 
to check your answer.</em>

"""

import numpy as np

# Simulate scores of 200 players
scores = np.random.normal(70, 15, 200)

# Count scores between 55 and 85
within_one_std = np.sum((scores >= 55) & (scores <= 85))

print("Total players:", len(scores))
print("Players scoring between 55 and 85:", within_one_std)
print("Percentage:", (within_one_std / 200) * 100, "%")

print("According to the 68-95-99 rule, approximately 68%")
print("of players are expected to score within one standard deviation.")
```
