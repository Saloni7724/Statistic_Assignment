```python
"""  
1. Probability of Getting Heads

Write a Python function calculate_probability(event_count, total_outcomes) that returns 
the probability of an event, and use it to find the probability of getting 'Heads' when
 tossing a coin.

"""
def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes

heads_probability = calculate_probability(1, 2)

print("Probability of getting Heads:", heads_probability)
```

```python
"""  
2. Experimental Probability of Getting a 6

Simulate rolling a six-sided dice 100 times using Python's random module, count how many 
times a '6' appears, and calculate the experimental probability of getting a '6'.<br><br>
<em><strong>Hint:</strong> Use random.randint(1,6) in a loop and keep a counter for '6'.
</em>

"""
import random

rolls = 100
six_count = 0

for i in range(rolls):
    result = random.randint(1, 6)
    
    if result == 6:
        six_count += 1

experimental_probability = six_count / rolls

print("Number of times 6 appeared:", six_count)
print("Experimental probability of getting 6:", experimental_probability)
```

```python
""" 
3. Probability of Picking a Pop Song

Given an array of genres from your last 20 Spotify songs (like ['pop', 'rock', 'pop', 
'hiphop', ...]), write code to calculate the probability of randomly picking a 'pop' 
song from your recent history.

"""

genres = [
    'pop', 'rock', 'pop', 'hiphop', 'pop',
    'jazz', 'rock', 'pop', 'hiphop', 'pop',
    'rock', 'pop', 'jazz', 'pop', 'hiphop',
    'rock', 'pop', 'pop', 'jazz', 'rock'
]

pop_count = genres.count('pop')
total_songs = len(genres)

pop_probability = pop_count / total_songs

print("Total number of songs:", total_songs)
print("Number of Pop songs:", pop_count)
print("Probability of picking a Pop song:", pop_probability)
```

```python
"""
4. Probability of Ordering Both Food and Dessert

Write a function that, given two arrays — one for users who ordered food on Zomato and 
one for users who ordered dessert — calculates the probability that a randomly selected 
user ordered both food and dessert (i.e., intersection over total unique users).

"""

def probability_both(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    both_users = food_set.intersection(dessert_set)
    all_users = food_set.union(dessert_set)

    probability = len(both_users) / len(all_users)

    return probability, both_users, all_users


food_users = ['A', 'B', 'C', 'D', 'E', 'F']
dessert_users = ['C', 'D', 'E', 'G', 'H']

probability, both_users, all_users = probability_both(
    food_users,
    dessert_users
)

print("Users who ordered both food and dessert:", both_users)
print("Total unique users:", len(all_users))
print("Probability of ordering both:", probability)
```

```python
""" 5. Conditional Probability for Flipkart Recommendations

Explain with a short Python code example how conditional probability could be used to
improve recommendations in an app like Flipkart (e.g., probability of buying headphones
given user bought a phone).<br><br><em><strong>Hint:</strong> Use counts of users who 
bought both items vs total who bought a phone.</em>

"""

def conditional_probability(both_count, given_count):
    return both_count / given_count


phone_buyers = 100
phone_and_headphone_buyers = 30

probability_headphones_given_phone = conditional_probability(
    phone_and_headphone_buyers,
    phone_buyers
)

print(
    "Probability of buying headphones given a phone purchase:",
    probability_headphones_given_phone
)

print(
    "Percentage:",
    probability_headphones_given_phone * 100,
    "%"
)
```
