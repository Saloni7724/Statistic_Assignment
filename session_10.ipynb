```python
""" 
1.
Write a Python function bayes_posterior(prior, likelihood, evidence) that calculates and returns the posterior probability using Bayes theorem formula
for given values.
"""
def bayes_posterior(prior, likelihood, evidence):
    posterior = (likelihood * prior) / evidence
    return posterior

prior = 0.30
likelihood = 0.80
evidence = 0.50

posterior = bayes_posterior(prior, likelihood, evidence)

print("Posterior probability:", posterior)
```

```python
"""  
2.
Given the following scenario: On Swiggy, 10% of restaurants are 'premium', and 80% of premium restaurants offer a discount, while only 20% of regular
restaurants do. If you see a discount, use Bayes theorem in code to find the probability that the restaurant is premium.<br><br><em><strong>Hint:</strong>
Define prior, likelihood, and evidence in your code with comments.</em>
"""

def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence

prior_premium = 0.10


likelihood_discount_premium = 0.80


probability_discount = (
    0.80 * 0.10 +
    0.20 * 0.90
)

posterior = bayes_posterior(
    prior_premium,
    likelihood_discount_premium,
    probability_discount
)

print("Probability that the restaurant is premium given a discount:",
      posterior)

print("Percentage:", posterior * 100, "%")
```

```python
"""  
3.
Simulate a basic spam detection filter: Given a list of 20 email subjects (some spam, some not), and the probability that 'free' appears in spam vs non-spam
emails, write code to calculate the probability an email is spam if it contains 'free', using Bayes theorem.

"""
def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence


emails = [
    "free prize",
    "free money",
    "free gift",
    "meeting tomorrow",
    "free offer",
    "project report",
    "free reward",
    "hello friend",
    "free coupon",
    "job offer",
    
    "team meeting",
    "your order",
    "free update",
    "college notice",
    "bank statement",
    "family dinner",
    "appointment reminder",
    "free information",
    "project meeting",
    "your bill"
]

spam = emails[:10]

not_spam = emails[10:]

free_spam = sum("free" in email for email in spam)

free_not_spam = sum("free" in email for email in not_spam)

prior = len(spam) / len(emails)

likelihood = free_spam / len(spam)

evidence = (free_spam + free_not_spam) / len(emails)
answer = bayes_posterior(prior, likelihood, evidence)

print("Probability email is spam:", answer)
print("Percentage:", answer * 100, "%")
```
